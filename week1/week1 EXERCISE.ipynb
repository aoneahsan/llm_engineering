```python
# imports
import os
import json
from IPython.display import Markdown, display, update_display
from openai import OpenAI
from dotenv import load_dotenv

```

```python
load_dotenv(override=True)
api_key=os.getenv("OPENAI_API_KEY")

if api_key and api_key.startswith('sk-proj-') and len(api_key) > 10:
    print("API Key looks good so far")
else:
    print("Invalid API Key")
```

```python
MODEL="gpt-5-nano"
openai = OpenAI()
```

```python
# set up environment
system_prompt="""
you are a helpful assistant and an expert on python and coding
if you get asked, to explain a question related to python you do your best and explain with breif, yet structured and formatted answer
if user ask anything else, you reply "fuck off, not my domain"
"""
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
response = openai.chat.completions.create(model=MODEL, messages=[{'role': 'system', 'content': system_prompt}, {'role': 'user', 'content': question}])
```

```python
display(Markdown(response.choices[0].message.content))
```

```python
# Get gpt-4o-mini to answer, with streaming
response2 = openai.chat.completions.create(model=MODEL, messages=[{'role': 'system', 'content': system_prompt}, {'role': 'user', 'content': 'tell me a joke'}])
```
